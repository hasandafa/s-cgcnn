This notebook demonstrates how to fetch AlₓGa₁₋ₓAs structure data from the Materials Project API using the `MPFetcher` class.

## Objectives
1. ✅ Initialize `MPFetcher` with API key
2. ✅ Fetch GaAs (x=0.0) and AlAs (x=1.0) structures
3. ✅ Display structure properties
4. ✅ Compare with literature data

## Setup

```python
# Import required modules
import sys
from pathlib import Path

# Add src to path
sys.path.insert(0, str(Path.cwd().parent))

# Import s-CGCNN modules
from src.data_acquisition import MPFetcher
from src.utils import constants
from src.utils.logger_config import setup_logger

# Setup logger
logger = setup_logger('data_fetching_demo')
logger.info("Data fetching notebook started")

print("✅ Imports successful!")
```

## 1. Initialize MPFetcher

The `MPFetcher` class handles all interactions with the Materials Project API.

```python
# Initialize fetcher with API key from config
fetcher = MPFetcher(api_key_path='../config/mp_api_key.txt')

print(f"✅ MPFetcher initialized")
print(f"   MP-API version: {fetcher.get_api_version()}")
```

## 2. Fetch GaAs Structure (x = 0.0)

GaAs is the starting composition in the AlₓGa₁₋ₓAs system.

```python
# Fetch GaAs from Materials Project
gaas_mpid = "mp-2534"  # Materials Project ID for GaAs

print(f"Fetching GaAs structure (ID: {gaas_mpid})...")
gaas_data = fetcher.fetch_structure(gaas_mpid)

# Display key properties
print("\n" + "=" * 60)
print("GaAs PROPERTIES (x = 0.0)")
print("=" * 60)
print(f"Formula: {gaas_data['formula']}")
print(f"Space Group: {gaas_data['space_group']}")
print(f"Lattice Constant: {gaas_data['lattice_constant']:.4f} Å")
print(f"Band Gap: {gaas_data['band_gap']:.4f} eV ({gaas_data['band_gap_type']})")
print(f"Density: {gaas_data['density']:.4f} g/cm³")
print("=" * 60)
```

## 3. Fetch AlAs Structure (x = 1.0)

AlAs is the endpoint composition.

```python
# Fetch AlAs from Materials Project
alas_mpid = "mp-2172"  # Materials Project ID for AlAs

print(f"Fetching AlAs structure (ID: {alas_mpid})...")
alas_data = fetcher.fetch_structure(alas_mpid)

# Display key properties
print("\n" + "=" * 60)
print("AlAs PROPERTIES (x = 1.0)")
print("=" * 60)
print(f"Formula: {alas_data['formula']}")
print(f"Space Group: {alas_data['space_group']}")
print(f"Lattice Constant: {alas_data['lattice_constant']:.4f} Å")
print(f"Band Gap: {alas_data['band_gap']:.4f} eV ({alas_data['band_gap_type']})")
print(f"Density: {alas_data['density']:.4f} g/cm³")
print("=" * 60)
```

## 4. Compare with Literature Data

Let's compare MP-API data with literature values from ioffe.ru.

```python
import pandas as pd

# Create comparison table
comparison_data = {
    'Property': ['Lattice (Å)', 'Band Gap (eV)', 'Density (g/cm³)'],
    'GaAs (Lit)': [constants.LATTICE_GAAS, constants.BAND_GAP_GAAS, constants.DENSITY_GAAS],
    'GaAs (MP)': [gaas_data['lattice_constant'], gaas_data['band_gap'], gaas_data['density']],
    'AlAs (Lit)': [constants.LATTICE_ALAS, constants.BAND_GAP_ALAS, constants.DENSITY_ALAS],
    'AlAs (MP)': [alas_data['lattice_constant'], alas_data['band_gap'], alas_data['density']]
}

df_comparison = pd.DataFrame(comparison_data)

# Calculate differences
df_comparison['GaAs Δ%'] = ((df_comparison['GaAs (MP)'] - df_comparison['GaAs (Lit)']) / 
                             df_comparison['GaAs (Lit)'] * 100)
df_comparison['AlAs Δ%'] = ((df_comparison['AlAs (MP)'] - df_comparison['AlAs (Lit)']) / 
                             df_comparison['AlAs (Lit)'] * 100)

print("\nCOMPARISON: Literature vs Materials Project")
print("=" * 80)
print(df_comparison.to_string(index=False))
print("=" * 80)

# Analysis
print("\n📊 ANALYSIS:")
print(f"   • GaAs band gap difference: {df_comparison.iloc[1]['GaAs Δ%']:.2f}%")
print(f"   • AlAs band gap difference: {df_comparison.iloc[1]['AlAs Δ%']:.2f}%")
print(f"   • Band gap correction needed for DFT underestimation")
```

## 5. Visualize Structure

Display the crystal structure using pymatgen's visualization.

```python
from pymatgen.core import Structure

# Get Structure objects
gaas_structure = Structure.from_dict(gaas_data['structure'])
alas_structure = Structure.from_dict(alas_data['structure'])

print("GaAs Structure:")
print(gaas_structure)

print("\nAlAs Structure:")
print(alas_structure)

print("\n✅ Structures loaded successfully!")
print("   For interactive 3D viewing, use the Structure Viewer app:")
print("   python apps/structure_viewer_app.py")
```
